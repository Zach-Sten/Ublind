```python
# Import packages:
import ublind as ub
import scanpy as sc
import numpy as np
import matplotlib.pyplot as plt
from IPython.display import Audio
```

# Embeddings to music! 

```python
# load in adata:
adata = sc.read_h5ad('example.h5ad')
```

```python
# access the different instruments and cheatsheets using the helper functions:
ub.hlp.instruments()   # prints friendly table with aliases
ub.hlp.scales()        # prints scales with notes and vibes
ub.hlp.gm_table()      # full 128 GM instruments
ub.hlp.quickstart()    # usage cheat sheet
```

## Listen across UMAP or your emebdding space of choice.

```python
# What does it sound like as we sweep UMAP space?
ub.pp.compose(adata, embedding="X_umap", time=10, scale="pentatonic",
                 counterpoint = False, # This vaies the axis so that low notes on the 1st axis start of the left but low notes on the 2nd axis start at the top.
                 dim_inst_map={0: "piano", 1: "piano"}, subsample = 5000)
ub.tl.render(adata, "sweep.wav") # can change renderer to 'synth', 'midi', 'fluidsynth' but we highly recommend not using synth
Audio("sweep.wav")
```

```python
# See what an orchestra of sounds sounds like across many dimensions in PC space! 
ub.pp.compose(adata, embedding="X_pca", time=10, scale="pentatonic", subsample = 5000)
ub.tl.render(adata, "sweep_pca.wav", renderer = 'synth') # can change renderer to 'synth', 'midi', 'fluidsynth'
Audio("sweep_pca.wav")
```

```python
# Animated UMAP with audio... work in progress.
ub.pl.sweep(adata, color_by="Cell_Annotation_Mid", legend_loc="on data")
```

## Listen by cluster:

```python
# Optional play one cluster a time using various methods for organizing them.
ub.pp.compose_clusters(adata, embedding="X_umap", cluster_key="Cell_Annotation_Mid", counterpoint = False,
    time=10, order="largest", scale="pentatonic", subsample = 5000,
    dim_inst_map={0: "piano", 1: "cello"})
ub.tl.render(adata, "clusters.wav")
Audio("clusters.wav")
```

```python
# Animated UMAP with audio... work in progress. no
ub.pl.sweep_clusters(adata, legend_loc="on data")
```

## If spatial coordinates are available, listen to your clusters across space:

```python
# Sweep both axes simultaneously (piano left→right, cello bottom→top)
ub.pp.compose_clusters(adata, embedding="spatial", cluster_key="Cell_Annotation_Mid",
    time=10, order="alphabetical", scale="pentatonic",
    dim_inst_map={0: "piano", 1: "cello"})
ub.tl.render(adata, "clusters.wav")
```

```python
# Play by spatial position! 
ub.pl.sweep_spatial(adata, spatial_key="spatial", sweep_axis="both", subsample = None, # Better if you just keep the full structure.
                    color_by="Cell_Annotation_Mid", legend_loc="on data")
```
