## Scales & Instruments

### Scales

Set `scale=` in `ub.pp.preprocess()`. Set `root=` to transpose (0=C, 2=D, 4=E, 5=F, 7=G, 9=A, 11=B).

- `"pentatonic"` — C D E G A — always sounds good
- `"major"` — happy, bright
- `"minor"` — sad, dark
- `"blues"` — jazzy, soulful
- `"dorian"` — jazzy minor
- `"mixolydian"` — bluesy major
- `"whole_tone"` — dreamy, floating
- `"chromatic"` — all 12 semitones, atonal

### Instruments

Set `dim_inst_map=` in `ub.pp.preprocess()` — e.g. `{0: "piano", 1: "cello"}`.

- `"piano"` — Acoustic Grand Piano (GM 0)
- `"guitar"` — Acoustic Guitar (GM 25)
- `"bass"` — Electric Bass (GM 33)
- `"violin"` — Violin (GM 40)
- `"viola"` — Viola (GM 41)
- `"cello"` — Cello (GM 42)
- `"harp"` — Orchestral Harp (GM 46)
- `"trumpet"` — Trumpet (GM 56)
- `"trombone"` — Trombone (GM 57)
- `"horn"` — French Horn (GM 60)
- `"sax"` — Tenor Sax (GM 66)
- `"clarinet"` — Clarinet (GM 71)
- `"flute"` — Flute (GM 73)
- `"vibraphone"` — Vibraphone (GM 11)

Or pass any GM program number directly: `{0: 42}` for Cello.

## Inspect Data

```python
# Load in example adata
adata = sc.read_h5ad('example.h5ad')
```

## Embeddings to music! 
The first dimension of the embedding becomes **time** (left → right),
and each remaining dimension becomes a separate **instrument voice**.

```python
# Basic umap made on a pentatonic scale
ub.pp.preprocess(adata, embedding="X_scVI", time=10, dim_inst_map={0: "harp"}, scale="pentatonic", subsample=5000,)
ub.tl.render(adata, "output.wav", renderer="synth", reverb=0.2) # Choose from synth, midi, or fluidsynth if installed.
Audio("output.wav")
```
