## Scales & Instruments

### Scales

Set `scale=` in `ub.pp.preprocess()`. Set `root=` to transpose (0=C, 2=D, 4=E, 5=F, 7=G, 9=A, 11=B).

- `"pentatonic"` — C D E G A — always sounds good
- `"major"` — happy, bright
- `"minor"` — sad, dark
- `"blues"` — jazzy, soulful
- `"dorian"` — jazzy minor
- `"mixolydian"` — bluesy major
- `"whole_tone"` — dreamy, floating
- `"chromatic"` — all 12 semitones, atonal

### Instruments

Set `dim_inst_map=` in `ub.pp.preprocess()` — e.g. `{0: "piano", 1: "cello"}`.

- `"piano"` — Acoustic Grand Piano (GM 0)
- `"guitar"` — Acoustic Guitar (GM 25)
- `"bass"` — Electric Bass (GM 33)
- `"violin"` — Violin (GM 40)
- `"viola"` — Viola (GM 41)
- `"cello"` — Cello (GM 42)
- `"harp"` — Orchestral Harp (GM 46)
- `"trumpet"` — Trumpet (GM 56)
- `"trombone"` — Trombone (GM 57)
- `"horn"` — French Horn (GM 60)
- `"sax"` — Tenor Sax (GM 66)
- `"clarinet"` — Clarinet (GM 71)
- `"flute"` — Flute (GM 73)
- `"vibraphone"` — Vibraphone (GM 11)

Or pass any GM program number directly: `{0: 42}` for Cello.

## Embeddings to music! 
The first dimension of the embedding becomes **time** (left → right),
and each remaining dimension becomes a separate **instrument voice**.

```python
adata = sc.read_h5ad('example.h5ad')
```

## Listen across UMAP or your emebdding space of choice.

```python
# Preprocess as you sweep across the UMAP
ub.pp.preprocess(adata, embedding="X_umap", time=10, scale="pentatonic",
    dim_inst_map={0: "piano", 1: "flute"}, subsample = 5000)
ub.tl.render(adata, "sweep.wav")
Audio("sweep.wav")
```

```python
# Animated UMAP with audio... work in progress.
ub.pl.sweep(adata, color_by="Cell_Annotation_Mid", legend_loc="on data")
```

## Listen by cluster:

```python
# Optional play one cluster a time using various methods for organizing them.
ub.pp.preprocess_clusters(adata, embedding="X_umap", cluster_key="Cell_Annotation_Mid",
    time=10, order="largest", scale="pentatonic",
    dim_inst_map={0: "piano", 1: "cello"})
ub.tl.render(adata, "clusters.wav")
Audio("clusters.wav")
```

```python
# Animated UMAP with audio... work in progress. no
ub.pl.sweep_clusters(adata, legend_loc="on data")
```
